##Adding usefull columns to the dataset

```python
## Adding Revenue

df["revenue"]=df["transaction_qty"]*df["unit_price" ]
df.head(10)
```

```python
#Convert columns to appropriate data types
df["transaction_date"]=pd.to_datetime(df["transaction_date"],errors="coerce")
df["transaction_time"]=pd.to_datetime(df["transaction_time"],errors="coerce")
```

```python
##checking the data types transformation  

df.dtypes
```

#Vizualization

```python
import matplotlib.pyplot as plt
```

```python
sales_by_date = df.groupby("transaction_date")["revenue"].sum()
sales_by_date.plot.line()
```

```python
##Another way to plot the graph with more details

plt.figure(figsize=(10,5))
plt.plot(sales_by_date.index,sales_by_date.values)
plt.title("Sales by Date")
plt.xlabel("month_name")
plt.xticks(rotation=45)
plt.ylabel("Revenue")
plt.tight_layout
plt.show()
```

```python
##Bar Graph: Sales by Store Location

sales_by_store_location = df.groupby("store_location")["revenue"].sum()

plt.figure(figsize=(7,5))
plt.bar(sales_by_store_location.index,sales_by_store_location.values)
plt.title("Sales by Store Location")
plt.xlabel("Store Location")
plt.ylabel("Revenue")
plt.tight_layout
plt.show()
```
